
## Using `matplotlib` in Julia


Load package(s):

```julia

using PythonPlot
```


Define a function to demo the plot: $f(x) = \sin(3x + 4\cos(2x))$

```julia

f(x) = sin(3*x + 4*cos(2*x));
```


Calculate data points:

```julia

x = range(0.0, 2*pi, 1000)
y = f.(x);
```


Plot the graph. Provide a grid, axis labels, title.

```julia

plot(x, y, color="blue", linewidth=1.5, linestyle="-")
grid(true)
xlabel("x")
ylabel("f(x)")
title("A sinusoidally modulated sinusoid");
```
